COMMENTAIRES MESURES : 
1) iD vs VI
Le courant reste presque nul tant que l’entrée est trop faible. 
Dès qu’on dépasse environ 0,6 – 0,7 V, la diode s’allume et le courant grimpe très vite. 

2) VO vs VI  
VO est la tension aux bornes de la résistance, donc VO = R × iD.
Tant que la diode ne conduit pas, iD ≈ 0, donc VO ≈ 0.
Quand VI dépasse le seuil de la diode, le courant augmente et VO augmente aussi, mais moins vite que VI, car une partie de la tension est “prise” par la diode (≈ 0,7 V).
On observe donc une montée progressive de VO, cohérente avec VO = VI – VD.

3) iD vs VD  
Pas de courant pour les tensions négatives, puis un seuil autour de 0,7 V, 
puis une montée exponentielle. En semilog, ça devient une droite, donc ça confirme le modèle exponentiel.


- Remplacer l'alimentation dc par un GBF et paramétrer le pour déterminer la résistance dynamique $r_d$ de la diode. Montrer que $r_d$ est égale en première approximation à $\frac{26 mV}{i_D}$. 

rd pour 3 pts de la caracteristique I_V: se placer sur un pt de polarisation et superposer un ac - extraire Di/Dv. n = 1 pour 1N4003
$$r_d = \frac{26mV}{i_d}$$

```python
#Mesure de rd
# injecter 100 mVpp au GBF autour d`un pt de polarisation en prendre 2 ou 3 (GBF + offset)
#rd = Vrms/Irms
#pts de polarisation [0.6V, 0.7V, 0.8V]
Vrms = 50e-3/np.sqrt(2)
Idrms = [0.025,0.028,0.029]   # composante ac mesurée au milliamperementre
rd = Vrms/np.array(Idrms)
print("rd = ", rd,"ohm")
Vd_DC = np.array([vd[4], vd[7], vd[10]])    # 0.529, 0.525, 0.585 prit des valeurs mesurées précédentes
Id_DC = np.array([Id[4], Id[7], Id[10]])     # 0.2mA, 0.9mA, 1.5mA pareil 
rf = Vd_DC / Id_DC
print("rf = ", rf,"ohm")
```

```python
# superposer mesure 
def plot2(vi, vo, vd, Id, data):
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(14,5))
    fig.suptitle('Superposition mesures et simulation')

    # vo = f(vi)
    ax1.plot(vi, vo, lw=3, label="mesures", color="red")
    ax1.plot(data["V1"], data["V(v0)"], lw=3, label="simulation", color="blue")
    ax1.set_xlabel("vi (V)")
    ax1.set_ylabel("vo (V)")
    ax1.grid()
    ax1.legend(frameon=False)

    # Id = f(vd)
    ax2.plot(vd, Id, lw=3, label="mesures", color="red")
    ax2.plot(data["V1"] - data["V(v0)"], data["I(D1)"], lw=3, label="simulation", color="blue")
    ax2.set_xlabel("vd (V)")
    ax2.set_ylabel("Id (A)")
    ax2.grid()
    ax2.legend(frameon=False)

    # Id = f(vi)
    ax3.plot(vi, Id, lw=3, label="mesures", color="red")
    ax3.plot(data["V1"], data["I(D1)"], lw=3, label="simulation", color="blue")
    ax3.set_xlabel("vi (V)")
    ax3.set_ylabel("Id (A)")
    ax3.grid()
    ax3.legend(frameon=False)

    plt.subplots_adjust(wspace=0.4, hspace=0.5)
    plt.show()

# appel de la fonction
plot2(vi, vo, vd, Id, data)
```

## Commentaires

Les simulations et les mesures sont bien identiques.

**Limiteur (a)**  
Les deux diodes montées en opposition limitent la tension dès que l’une d’elles devient passante, le signal est alors écrêté autour de +-0,7 V.

**Limiteur Zener (b)**  
Les deux diodes Zener montées en opposition limitent la tension de sortie. L’une conduit pour les crêtes positives et l’autre pour les crêtes négatives, ce qui maintient vo autour de Vz +-5 V.  

**Écrêteur (c)**  
La diode charge le condensateur quand l’entrée monte, puis il se décharge dans RL : du coup, les niveaux hauts penchent vers le bas et, comme le condensateur ne retrouve jamais sa charge complète, tout le signal décroît légèrement au fil des périodes.

**Doubleur de tension (d)**  
On voit que la sortie vo descend par paliers à chaque alternance, une diode conduit et les condensateurs se chargent ou se déchargent, ce qui fait perdre un peu de tension. Comme les condensateurs ne récupèrent jamais toute leur charge et que RL consomme, la sortie baisse progressivement jusqu’à se stabiliser proche de 0 .

## Questions
Oui, on peut modifier ces circuits :

- Pour limiter à +- 1,4 V dans le circuit limiteur, il faut mettre deux diodes en série au lieu d'une entre Vi et Vo, leurs chutes de tension vont alors s'additionner (+0,7V + 0,7V) ce qui donne environ 1,4 V de seuil et le même procédé dans l'autre sens pour l'avoir en négatif.

- Pour le montage écreteur (c), on peut obtenir un écrêtage négatif en retournant la diode vers la masse, ce qui ferait charger le condensateur dans l’autre sens et donc écrêter la partie négative du signal.

- Et pour le doubleur de tension, il faut inverser les diodes et la polarité du condensateur pour obtenir une sortie positive. Actuellement la sortie est naturellement négative parce que les diodes et les condensateurs sont orientés pour charger la tension vers le bas.

En somme, tout se joue sur l’orientation des diodes et la façon dont elles imposent leur chute de tension.

*Réponse question :*

**Déterminer la tension de seuil de Vgs, c’est‑à‑dire Vtn, pour laquelle le courant augmente :**

La tension de seuil est environ 2 V, c’est à ce moment que le courant Id commence réellement à augmenter (graphique de gauche).

**Quelles informations êtes‑vous susceptibles d’extraire de ces caractéristiques de sortie :**

Quand on trace Id en fonction de VDS pour plusieurs valeurs de VG, on voit d’abord la zone linéaire puis la zone de saturation.

On le voit clairement car au début la courbe monte presque droit, ce qui montre que le courant augmente fortement avec VDS, puis elle commence à se courber et à devenir presque plate, signe que le transistor ne laisse plus passer beaucoup plus de courant même si on augmente VDS.

On remarque aussi qu’augmenter VG fait monter toutes les courbes vers le haut. Cela se voit directement car pour une même valeur de VDS, le courant Id est plus élevé quand VG est plus grand.

Cela montre que la grille contrôle la conduction du canal et qu’un VG plus élevé ouvre davantage le transistor.

```python
#Traitement des simulations -  extraire k_n, V_tn et 1/Va = lambda
Vgs = a["VGS"].values
Id  = a["Id(M1)"].values
#print(a)

# Determiner Vt et kn prendre les données pour id vs. vgs  pour 1 Vds assurant la saturation  
# d'apres le cours en saturation (Vds >= Vgs - Vtn) on a :
# Id = (1/2)*kn*(Vgs-Vtn)^2  =>  sqrt(Id) = sqrt(kn/2) * (Vgs - Vtn)  => c'est une droite
# je selectionne donc uniquement les points en saturation
Vgs_sat     = Vgs[Vgs >= 2.2]
Id_sat      = Id[Vgs >= 2.2]
sqrt_Id_sat = np.sqrt(Id_sat)   # sqrt(Id) est lineaire en Vgs en saturation

# np.polyfit trouve la droite y = p[0]*x + p[1] qui passe au mieux par mes points
p   = np.polyfit(Vgs_sat, sqrt_Id_sat, 1)
# p[0] = pente = sqrt(kn/2)
# p[1] = ordonnee a l'origine = -sqrt(kn/2)*Vtn

# quand sqrt(Id) = 0  =>  Vgs = Vtn  =>  Vtn = -p[1]/p[0]
Vtn = -p[1] / p[0]

# pente = sqrt(kn/2)  =>  kn = 2*pente^2
kn  = 2 * p[0]**2

print(f"Vtn = {Vtn:.4f} V")
print(f"kn  = {kn*1e6:.2f} uA/V^2")

#tracer sqrt(Id) vs. vgs
fig, ax1 = plt.subplots(1, 1, figsize=(6,5))
fig.suptitle('Extraction de Vtn et kn')
ax1.plot(Vgs, np.sqrt(Id), 'b.')
ax1.set_xlabel('$V_{GS}$ (V)')
ax1.set_ylabel('$\sqrt{I_d}$')
ax1.grid()
plt.show()
```
